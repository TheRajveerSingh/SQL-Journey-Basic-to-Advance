```python
%%sql
CREATE TABLE Employee (
    Eid INT PRIMARY KEY,
    Name VARCHAR(100),
    Deep_Id INT,
    Qualification VARCHAR(50),
    Sex CHAR(1)
);
INSERT INTO Employee (Eid, Name, Deep_Id, Qualification, Sex) VALUES
(1, 'Deepali Gupta', 101, 'MCA', 'F'),
(2, 'Rajat Tyagi', 101, 'BCA', 'M'),
(3, 'Hari Mohan', 102, 'B.A.', 'M'),
(4, 'Harry', 102, 'M.A.', 'M'),
(5, 'Sunit Mittal', 103, 'B.Tech', 'M'),
(6, 'Jyoti', 101, 'M.Tech', 'F');
```

```python
%%sql
CREATE TABLE Salary (
    Eid INT PRIMARY KEY,
    Basic INT,
    DA INT,
    HRA INT,
    Bonus INT,
    FOREIGN KEY (Eid) REFERENCES Employee(Eid)
);
INSERT INTO Salary (Eid, Basic, DA, HRA, Bonus) VALUES
(1, 6000, 2000, 2300, 200),
(2, 2000, 300, 300, 30),
(3, 1000, 300, 300, 40),
(4, 1500, 390, 490, 30),
(5, 8000, 900, 400, 80),
(6, 10000, 300, 490, 89);
```

```python
%%sql
# To display the frequency of employees department wise
SELECT DISTINCT Deep_ID as Department, COUNT(*) FROM Employee GROUP BY Deep_ID;
# Even COUNT(NAME) would have worked because its just counting the unique rows, and again, no need for adding DISTINCT.
```

```python
%%sql
#To list the names of those employees only whose name starts with 'H'.
SELECT Name FROM Employee WHERE Name like 'H%';
```

```python
%%sql
#To add a new column in salary table. The coloumn name is Total Sal.
ALTER TABLE Salary ADD Total_Sal int;
#There cannot be space between Total and Sal. Also, its important to add the type of the coloumn while adding it.
```

```python
%%sql
#To store the corresponding values in Total Sal coloumn.
#UPDATE TABLE_NAME to set values in new coloumns
UPDATE Salary SET Total_Sal = Basic + DA+ HRA + Bonus ;
```
