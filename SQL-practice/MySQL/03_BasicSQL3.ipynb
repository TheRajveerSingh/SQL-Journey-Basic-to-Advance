```python
%%sql
CREATE TABLE TRADERS (
TCODE VARCHAR(10) PRIMARY KEY, TNAME VARCHAR(100) NOT NULL, CITY VARCHAR(50) NOT NULL);
```

```python
%%sql
INSERT INTO TRADERS (TCODE, TNAME, CITY) VALUES
('T01', 'ELECTRONIC SALES', 'MUMBAI'),
('T03', 'BUSY STORE CORP', 'DELHI'),
('T02', 'DISP HOUSE INC', 'CHENNAI');
```

```python
%%sql
CREATE TABLE Items (
CODE INT PRIMARY KEY, INAME VARCHAR(100) NOT NULL, QTY INT NOT NULL,
PRICE DECIMAL(10, 2) NOT NULL, COMPANY VARCHAR(100) NOT NULL,
TCODE VARCHAR(10), FOREIGN KEY (TCODE) REFERENCES TRADERS(TCODE));
```

```python
%%sql
#Display the TCODE, TNAME, INAME FROM THE TABLES
SELECT T.TCODE, T.TNAME, I.INAME FROM TRADERS T, Items I WHERE T.TCODE=I.TCODE;
```

```python
%%sql
#Display the INAME, TCODE, TNAME where quantity is less than 100.
SELECT I.INAME, T.TCODE, T.TNAME FROM TRADERS T, Items I WHERE T.TCODE=I.TCODE and I.QTY<100;
```

```python
%%sql
#Display TNAME and the Total No. of Items that each trader bought.
SELECT DISTINCT T.TNAME, SUM(QTY) FROM TRADERS T, Items I WHERE T.TCODE=I.TCODE GROUP BY T.TNAME;
```

```python
%%sql
#Display the total number of companies each trader had bought.
SELECT DISTINCT T.TNAME, COUNT(I.COMPANY) FROM TRADERS T, Items I WHERE T.TCODE=I.TCODE GROUP BY T.TNAME;
```
