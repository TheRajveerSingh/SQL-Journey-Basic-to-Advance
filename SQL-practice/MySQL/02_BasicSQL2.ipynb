```python
%%sql
CREATE TABLE CARDEN (
    Ccode INT PRIMARY KEY,
    CarName VARCHAR(50),
    Make VARCHAR(50),
    Color VARCHAR(20),
    Capacity INT,
    Charges INT
);
```

```python
%%sql
SELECT * FROM CARDEN;
```

```python
%%sql
#To display all the information about the Suzuki Cars.
SELECT * FROM Carden WHERE Make = "Suzuki";
```

```python
%%sql
#To display the Ccode, CarName, and Color in descending order of their capacity.
SELECT Ccode, CarName, Color FROM Carden ORDER BY Capacity desc;
```

```python
%%sql
#To display the names of all red colored cars that can accomodate 4 people.
SELECT CarName FROM Carden WHERE Color = "Red" and Capacity>3;
#Okay here, notice how we try finding "Red" and it works, even though the table had "RED".
```

```python
%%sql
#To display the total number of models
SELECT COUNT(DISTINCT Make) as Total_Number_of_Models FROM Carden;
```

```python
%%sql
#To display the information of all cars where the name starts with the letter 'I' in ascending order of CarName
SELECT * FROM Carden WHERE CarName LIKE 'I%' ORDER BY CarName asc;
```

```python
%%sql
#To display the names of cars where the capacity is more than 4 and Ccode is more than 505.
SELECT CarName FROM Carden WHERE Capacity>4 and Ccode>505;
```

```python
%%sql
#To display no.of cars of each make
SELECT DISTINCT MAKE, COUNT(*) FROM Carden GROUP BY Make;
```

```python
%%sql
#Even if we don't mention distinct, GROUP BY clause automatically takes unique categories.
SELECT MAKE, COUNT(*) FROM Carden GROUP BY Make;
```
