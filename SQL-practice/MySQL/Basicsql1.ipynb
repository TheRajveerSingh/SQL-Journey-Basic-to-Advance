```python
%%sql
CREATE TABLE Student (
No INT PRIMARY KEY, Name VARCHAR(50), Age INT, Department VARCHAR(50), Date_of_adm DATE,
Fee INT,  Gender CHAR(1));
```

```python
%%sql
#To display the table
SELECT * FROM Student;
```

```python
%%sql
#To show information of all students from History department
SELECT * FROM Student WHERE Department = "History";
```

```python
%%sql
#To list the names of all female Students who are in Hindi Department
SELECT Name FROM Student WHERE Department = "Hindi" and Gender = "F";
```

```python
%%sql
#To display the names of all students with their date of admission in ascending order.
SELECT Name, Date_of_adm FROM Student ORDER BY Date_of_adm ASC;
```

```python
%%sql
#To count the number of students whose Date of Admission was in 1998.
SELECT COUNT(*) as Number_of_Students FROM Student WHERE Date_of_adm LIKE '1998%';
```

```python
%%sql
#To display the names of all departments.
SELECT DISTINCT Department as Departments FROM Student;
#Distinct helps choose all unique Departments from the table, not repeating them.
```
