```python
%%sql
CREATE TABLE SHOP (
    ID VARCHAR(10) PRIMARY KEY,
    SName VARCHAR(100) NOT NULL,
    Area VARCHAR(50) NOT NULL
);
```

```python
%%sql
CREATE TABLE ACCESSORIES (
    ID VARCHAR(10) PRIMARY KEY,
    IName VARCHAR(100) NOT NULL,
    Price DECIMAL(10, 2) NOT NULL,
    SNo VARCHAR(10),
    FOREIGN KEY (SNo) REFERENCES SHOP(ID)
);
INSERT INTO ACCESSORIES (ID, IName, Price, SNo) VALUES
('A01', 'Mother Board', 12000.00, 'S01'),
('A02', 'Hard Disk', 5000.00, 'S01'),
('A03', 'Keyboard', 500.00, 'S02'),
('A04', 'Mouse', 300.00, 'S01'),
('A05', 'Mother Board', 13000.00, 'S02'),
('A06', 'Keyboard', 400.00, 'S03'),
('A07', 'LCD', 6000.00, 'S04'),
('A08', 'LCD', 5500.00, 'S05'),
('A09', 'Mouse', 350.00, 'S05'),
('A10', 'Hard Disk', 4500.00, 'S03');
```

```python
%%sql
#To display Name and Price of the Accesories in ascending order of its price.
SELECT A.INAME, A.PRICE FROM ACCESSORIES A ORDER BY Price asc;
```

```python
%%sql
#To display the ID and SName of all Shops located in Nehru Place
SELECT ID, SName FROM SHOP WHERE Area = "Nehru Place";
```

```python
%%sql
#To display the minimum and maximum price of each name of Accessories
SELECT DISTINCT IName, MIN(Price), MAX(Price) FROM ACCESSORIES GROUP BY IName;
```

```python
%%sql
#To display the Name, Price of all accessories and their respectful SName where they are available
SELECT A.IName, A.Price, S.Area FROM ACCESSORIES A, SHOP S WHERE A.SNo = S.ID;
```

```python
%%sql
#TO display the total number of items that each shop has, ,along with the shop name.
SELECT SName, COUNT(INAME) FROM ACCESSORIES A, SHOP S WHERE S.ID = A.Sno GROUP BY SName;
#As you can see here, No need to put A. or S. If the coloumns in both tables are different. And also, Since we;ve added GROUP BY CLAUSE, no need to write DISTINCT SName.
```

```python
%%sql
#To add a coloumn named 'Discounted Price' into the Accessories Table.
#SQL command to add a coloumn into a table
ALTER TABLE Accessories ADD `Discounted Price` INT;
```

```python
%%sql
#To add values to the coloumn 'Discounted Price' by giving 12% discount of the Price.
UPDATE accessories SET `Discounted Price` = Price - (Price * 0.12);
```

Okay, Now we'll see how a same command in SQL can be written both simply, and using JOINS:

```python
%%sql
select IName, Price, SName from Shop S, Accessories A where S.ID=A.SNO;
```

```python
%%sql
select IName, Price, SName from Shop S join Accessories A on S.ID = A.SNO;
```

```python
%%sql
#SQL command to delete a coloumn into a table
ALTER TABLE Accessories DROP COLUMN `Discounted Price`;
```
